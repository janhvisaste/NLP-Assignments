```python
import nltk
from nltk.tokenize import WhitespaceTokenizer, wordpunct_tokenize
from nltk.tokenize import TreebankWordTokenizer, TweetTokenizer
from nltk.tokenize import MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Download required resources (run once)
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = """
OMG!!! I can't believe this 🤯🔥
AI is evolving sooo fast!!! Check this out: https://openai.com
#AI #DeepLearning #Future
"""
```

White Space Tokenization - Splits text wherever whitespace appears.

```python
whitespace_tokenizer = WhitespaceTokenizer()
tokens_ws = whitespace_tokenizer.tokenize(text)
print("Whitespace Tokenization:\n", tokens_ws)
```

Punctuation-Based Tokenization : Splits words and punctuation separately.

```python
tokens_punct = wordpunct_tokenize(text)
print("Punctuation-based Tokenization:\n", tokens_punct)
```

Treebank Tokenization : Handles contractions properly (used in Penn Tree Banks)

```python
treebank_tokenizer = TreebankWordTokenizer()
tokens_treebank = treebank_tokenizer.tokenize(text)
print("Treebank Tokenization:\n", tokens_treebank)
```

Tweet Tokenization : Specially designed for Social Media Text
Handles:
*   Emojis
*   HashTags
*   URLs

```python
tweet_tokenizer = TweetTokenizer()
tokens_tweet = tweet_tokenizer.tokenize(text)
print("Tweet Tokenization:\n", tokens_tweet)
```

MWE (Multi-Word Expression) Tokenization : Combines predefined word pairs.

```python
mwe_tokenizer = MWETokenizer([('Machine', 'Learning')])
text2 = "I love Machine Learning and Natural Language Processing"
tokens_mwe = mwe_tokenizer.tokenize(text2.split())
print("MWE Tokenization:\n", tokens_mwe)
```

Stemming : Stemming reduces words to their root form (not always real words).

```python
# Porter Stemmer
porter = PorterStemmer()

words = ["running", "runs", "easily", "fairly"]
porter_stems = [porter.stem(word) for word in words]

print("Porter Stemming:\n", porter_stems)
```

```python
# Snowball Stemmer : more advanced than porter
snowball = SnowballStemmer("english")
snowball_stems = [snowball.stem(word) for word in words]

print("Snowball Stemming:\n", snowball_stems)
```

Lemmatization : Lemmatization returns meaningful base words using vocabulary + POS.

```python
#WordNet Lemmetizer
lemmatizer = WordNetLemmatizer()

words = ["running", "better", "flies"]

# Default (noun)
lemmatized_default = [lemmatizer.lemmatize(word) for word in words]
print("Lemmatization (default noun):\n", lemmatized_default)

# With POS tagging
lemmatized_verbs = [lemmatizer.lemmatize(word, pos='v') for word in words]
print("Lemmatization (verb):\n", lemmatized_verbs)
```
